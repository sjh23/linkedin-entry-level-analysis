```python
import pandas as pd
from pathlib import Path

DATA = Path("../data")

# 1) postings.csv: 5행만 읽어서 컬럼 확인
sample = pd.read_csv(DATA / "postings.csv", nrows=5)
print(sample.columns.tolist())
sample.T  # 세로로 뒤집어 보면 컬럼이 많아도 보기 편해요
```

```python
# 2) 나머지 테이블들의 컬럼 한 번에 보기
for f in sorted(DATA.rglob("*.csv")):
    if f.name == "postings.csv":
        continue
    df = pd.read_csv(f, nrows=3)
    print(f"{f.relative_to(DATA)}  →  {df.columns.tolist()}")
```

```python
# 3) skills 매핑 확인 (작아서 전체를 읽어도 돼요)
pd.read_csv(DATA / "mappings" / "skills.csv")
```

```python
cols = ["job_id", "title", "formatted_experience_level", "normalized_salary",
        "pay_period", "remote_allowed", "company_id", "formatted_work_type",
        "views", "applies"]
df = pd.read_csv(DATA / "postings.csv", usecols=cols)
print(df.shape)

# 컬럼별 결측 비율 (1.0 = 100% 비어 있음)
df.isna().mean().sort_values(ascending=False).round(3)
```

```python
# 가장 중요한 셀: 경력 레벨 분포
df["formatted_experience_level"].value_counts(dropna=False)
```
